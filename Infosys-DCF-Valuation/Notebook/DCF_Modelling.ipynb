```python
import pandas as pd
file_path = "Infosys.xlsx"
income = pd.read_excel(file_path, sheet_name="Income Statement")
cashflow = pd.read_excel(file_path, sheet_name="Cash Flow")
balance = pd.read_excel(file_path, sheet_name="Balance Sheet")
```

```python
income.set_index(income.columns[0], inplace=True)
cashflow.set_index(cashflow.columns[0], inplace=True)
balance.set_index(balance.columns[0], inplace=True)
print("Income Statement:")
print(income.head())

print("\nCash Flow:")
print(cashflow.head())
```

```python
income.index = income.index.str.strip()
cashflow.index = cashflow.index.str.strip()
balance.index = balance.index.str.strip()
```

```python
revenue = income.loc["Sales"]
ebit = income.loc["EBIT"]
cfo = cashflow.loc["Cash from Operating Activity"]
capex = cashflow.loc["CapEx (Fixed Assets Purchased)"]
print("Revenue:")
print(revenue)
print("\nEBIT:")
print(ebit)
print("\nCFO:")
print(cfo)
print("\nCapEx:")
print(capex)
```

```python
#calculating FCF
fcf = cfo + capex
print("Free Cash Flow:")
print(fcf)
```

```python
# Calculating year-on-year growth
growth = fcf.pct_change()
print("FCF Growth Rate:")
print(growth)
```

```python
filtered_growth = growth[(growth > -0.2) & (growth < 0.3)]
avg_growth = filtered_growth.mean()
print("Normalized Growth:", avg_growth)
years = 5
last_fcf = fcf.iloc[-1]
fcf_forecast = []
for i in range(1, years+1):
    future_fcf = last_fcf * (1 + avg_growth)**i
    fcf_forecast.append(future_fcf)
fcf_forecast = [float(x) for x in fcf_forecast]
print(fcf_forecast)
```

```python
import yfinance as yf

# Use explicit date range
stock = yf.download("INFY.NS", start="2018-01-01", end="2026-01-01")
market = yf.download("^NSEI", start="2018-01-01", end="2026-01-01")

print(stock.head())
print(stock.tail())
print(stock.index.min())
print(stock.index.max())
```

```python
stock_close = stock['Close']
market_close = market['Close']
```

```python
stock_returns = stock_close.pct_change()
market_returns = market_close.pct_change()

data = pd.concat([stock_returns, market_returns], axis=1).dropna()
data.columns = ['Stock', 'Market']
```

```python
import numpy as np

cov = np.cov(data['Stock'], data['Market'])[0][1]
var = np.var(data['Market'])
beta = cov / var
print("Real Beta:", beta)
```

```python
# Interest expense from Income Statement
interest = income.loc["Interest"].iloc[-1]

# Total debt from Balance Sheet
debt = balance.loc["Borrowings"].iloc[-1]

# Cost of Debt
cost_of_debt = interest / debt

print("Cost of Debt:", cost_of_debt)
```

```python
# Cost of Equity
rf = 0.07
rm = 0.12

cost_of_equity = rf + beta * (rm - rf)

# Capital structure
equity = balance.loc["Equity Capital"] + balance.loc["Reserves"]
debt = balance.loc["Borrowings"]

equity_latest = equity.iloc[-1]
debt_latest = debt.iloc[-1]

# Tax rate
tax_rate = income.loc["Tax %"].iloc[-1] / 100

# Total value
total_value = equity_latest + debt_latest

# WACC
wacc = (equity_latest / total_value) * cost_of_equity + \
       (debt_latest / total_value) * cost_of_debt * (1 - tax_rate)

print("WACC:", wacc)
```

```python
discounted_fcf = []

for i in range(len(fcf_forecast)):
    pv = fcf_forecast[i] / (1 + wacc)**(i+1)
    discounted_fcf.append(pv)

discounted_fcf = [float(x) for x in discounted_fcf]

print("Discounted FCF:")
print(discounted_fcf)
```

```python
g = 0.04  # long-term growth (4%)

terminal_value = fcf_forecast[-1] * (1 + g) / (wacc - g)

pv_terminal = terminal_value / (1 + wacc)**len(fcf_forecast)

print("Terminal Value:", terminal_value)
print("PV of Terminal Value:", pv_terminal)
```

```python
enterprise_value = sum(discounted_fcf) + pv_terminal

print("Enterprise Value:", enterprise_value)
```

```python
cash = balance.loc["Investments"].iloc[-1] + balance.loc["Other Assets"].iloc[-1]
debt = balance.loc["Borrowings"].iloc[-1]

equity_value = enterprise_value + cash - debt

print("Equity Value:", equity_value)
```

```python
shares = 2073   # Equity capital (₹ crore)

value_per_share = equity_value / shares

print("Intrinsic Value per Share:", value_per_share)
```

```python
wacc_range = [0.09, 0.10, 0.11, 0.12,0.13]
g_range = [0.03, 0.04, 0.05,0.06]
```

```python
sensitivity_table = []

for w in wacc_range:
    row = []
    for g in g_range:
        
        # Terminal value
        tv = fcf_forecast[-1] * (1 + g) / (w - g)
        
        # PV of terminal value
        pv_tv = tv / (1 + w)**len(fcf_forecast)
        
        # Discounted FCF
        pv_fcf = sum([fcf_forecast[i] / (1 + w)**(i+1) for i in range(len(fcf_forecast))])
        
        # Enterprise value
        ev = pv_fcf + pv_tv
        
        row.append(ev)
    
    sensitivity_table.append(row)
```

```python
import pandas as pd

sensitivity_df = pd.DataFrame(
    sensitivity_table,
    index=[f"{int(w*100)}%" for w in wacc_range],
    columns=[f"{int(g*100)}%" for g in g_range]
)

print(sensitivity_df)
```

```python
formatted_df = sensitivity_df.copy()

for col in formatted_df.columns:
    formatted_df[col] = formatted_df[col].apply(lambda x: f"{x:,.0f}")

print(formatted_df)
```

```python
sensitivity_df.index.name = "WACC (%)"
sensitivity_df.columns.name = "Terminal Growth Rate (%)"
styled_df = sensitivity_df.style \
    .format("{:,.0f}") \
    .background_gradient(cmap="RdYlGn") \
    .set_caption("DCF Sensitivity Analysis (Enterprise Value in ₹ Crore)") \
    .set_properties(**{'text-align': 'center'})

styled_df
```

```python
scenarios = {
    "Bear": {"wacc": 0.12, "g": 0.03},
    "Base": {"wacc": 0.10, "g": 0.04},
    "Bull": {"wacc": 0.09, "g": 0.05}
}
```

```python
scenario_results = {}

for name, vals in scenarios.items():
    w = vals["wacc"]
    g = vals["g"]

    # Terminal value
    tv = fcf_forecast[-1] * (1 + g) / (w - g)
    pv_tv = tv / (1 + w)**len(fcf_forecast)

    # Discounted FCF
    pv_fcf = sum([fcf_forecast[i] / (1 + w)**(i+1) for i in range(len(fcf_forecast))])

    # Enterprise Value
    ev = pv_fcf + pv_tv

    scenario_results[name] = ev

scenario_results
```

```python
scenario_df = pd.DataFrame.from_dict(scenario_results, orient='index', columns=['Enterprise Value'])

scenario_df
```

```python
scenario_df["Enterprise Value"] = scenario_df["Enterprise Value"].apply(lambda x: f"{x:,.0f}")

print(scenario_df)
```

```python
scenario_df_numeric = pd.DataFrame.from_dict(scenario_results, orient='index', columns=['EV'])

scenario_df_numeric["Equity Value"] = scenario_df_numeric["EV"] + cash - debt
scenario_df_numeric["Value per Share"] = scenario_df_numeric["Equity Value"] / shares

scenario_df_numeric
```

```python
final_df = pd.DataFrame.from_dict(scenario_results, orient='index', columns=['Enterprise Value'])

# Add Equity Value
final_df["Equity Value"] = final_df["Enterprise Value"] + cash - debt

# Add Value per Share
final_df["Value per Share"] = final_df["Equity Value"] / shares

def highlight_base(row):
    if row.name == 'Base':
        return ['background-color: yellow'] * len(row)
    return [''] * len(row)

styled_final = final_df.style \
    .set_caption("Scenario Valuation Summary (₹ Crore & Per Share)") \
    .background_gradient(cmap="RdYlGn", subset=["Enterprise Value", "Equity Value"]) \
    .apply(highlight_base, axis=1) \
    .format({
        "Enterprise Value": "{:,.0f}",
        "Equity Value": "{:,.0f}",
        "Value per Share": "{:,.0f}"
    }) \
    .set_properties(**{
        'text-align': 'center',
        'font-size': '12pt'
        
    }) \
    .set_table_styles([
    {'selector': 'th', 'props': [('border', '1px solid black')]},
    {'selector': 'td', 'props': [('border', '1px solid black')]}
])
styled_final
```

```python
DCF Valuation of Infosys
1. Objective

The objective of this project is to estimate the intrinsic value of Infosys using the Discounted Cash Flow (DCF) method. 
The model helps determine whether the stock is overvalued or undervalued based on future cash flow projections.

2. Data Collection

Financial data for Infosys was collected from Screener and Yahoo Finance. 
The data includes income statement, cash flow statement, and balance sheet items required for valuation.

3. Financial Analysis

The company’s financial performance was analyzed using key metrics such as revenue, operating profit, and net profit. 
This helps in understanding historical growth trends and profitability.

4. Free Cash Flow Calculation

Free Cash Flow (FCF) was calculated using cash from operating activities and capital expenditure. 
FCF represents the cash available to investors after accounting for investments in fixed assets.

Formula used:
FCF = Cash Flow from Operations + Capital Expenditure

5. Forecasting

Future free cash flows were projected for a period of 5 years using historical growth trends.
Growth rates were normalized to remove extreme fluctuations and ensure realistic projections.

6. WACC Calculation

The Weighted Average Cost of Capital (WACC) was calculated to discount future cash flows.
Cost of equity was estimated using the Capital Asset Pricing Model (CAPM), and cost of debt was derived from financial statement data.

7. Key Assumptions

The valuation is based on the following assumptions:

Risk-free rate: 7%
Market return: 12%
Beta: Calculated using market data
Terminal growth rate: 3%–5%
Forecast period: 5 years
WACC: Approximately 10–11%

8. DCF Valuation

Projected free cash flows were discounted using WACC to estimate their present value. 
A terminal value was calculated using the Gordon Growth Model to capture value beyond the forecast period.

9. Sensitivity Analysis

Sensitivity analysis was performed to evaluate how changes in WACC and terminal growth rate impact the valuation.
This helps assess the robustness and reliability of the model.

10. Scenario Analysis

Three scenarios were considered:

Bear case: Conservative assumptions
Base case: Most realistic assumptions
Bull case: Optimistic assumptions

This provides a range of possible valuations under different conditions.

11. Conclusion

The DCF model estimates a fair value of approximately ₹404 per share under the base case. 
The valuation ranges from ₹292 to ₹566 across different scenarios, highlighting the sensitivity of valuation to key assumptions.

12. Limitations

The model has certain limitations. 
The valuation is highly sensitive to assumptions such as growth rate and discount rate.
External market conditions and qualitative factors are not fully captured in the model.
```
